# Handling Imbalanced Dataset

##### 1. Up Sampling
##### 2. Down Sampling

```python
import numpy as np
import pandas as pd

# set the random seed for reproudcibility
np.random.seed(143)

# create a dataframe with two classed
n_samples = 1000
class_0_ratio = 0.9

n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
class_0 = pd.DataFrame({
    'feaure_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feaure_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target' : [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feaure_1' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'feaure_2' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'target' : [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsampled = resample(df_minority, replace=True, # sample will be added
        n_samples = len(df_majority), random_state = 42)
```

```python
df_minority_upsampled.shape
```

```python
final_df = pd.concat([df_minority_upsampled, df_majority]).reset_index(drop=True)
```

```python
len(final_df[final_df['target']==0])
```

```python
# downsample

df_majority_downsampled = resample(df_majority, replace = False, 
                              n_samples = len(df_minority), random_state = 42)
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority, df_majority_downsampled]).reset_index(drop=True)
df_downsampled.shape
```
