## Nominal/One Hot Encoding

#### Encoding is converting categorical data into a meaningful numerical value so that machine learning algorithm can understand and process it

#### One Hot Encoding creates a new column for every unique category and represent the presence of the cateogry with 1 and absence with 0

#### Red [0,0, 1]
#### Green [0, 1, 0]
#### Blue [1, 0, 0]

# Disadvantages

#### 1. Number of Features - for n number of unique categories it creates n number of columns 

#### 2. Sparse Matrix - the matrix contains 0's and 1's, leads to overfitting

```python
import seaborn as sns
dataset = sns.load_dataset('tips')
```

```python
sex_encoder = OneHotEncoder()
```

```python
sex_encoded = sex_encoder.fit_transform(dataset[['sex']]).toarray()
```

```python
sex_encoded_df = pd.DataFrame(sex_encoded, columns = sex_encoder.get_feature_names_out())
```

```python
sex_encoded_df
```

```python
smoker_encoder = OneHotEncoder()
```
