# SMOTE - Synthetic Minority Oversampling Technique

##### SMOTE (Synthetic Minority Over-sampling Technique) is an oversampling technique used to handle imbalanced datasets. Instead of simply duplicating minority class samples, SMOTE creates synthetic data points by interpolating between existing minority class data points. This helps generate more meaningful and diverse samples for the minority class.

```python
from sklearn.datasets import make_classification
```

```python
X, y  = make_classification(n_samples=1000, n_features = 2, n_clusters_per_class = 2, weights=[0.90], n_redundant = 0, random_state=12)
```

```python
import pandas as pd

df1 = pd.DataFrame(X, columns = ['f1', 'f2'])
df2 = pd.DataFrame(y, columns = ['target'])

df = pd.concat([df1, df2], axis = 1)
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['f1'], df['f2'], c = df['target'])
```

```python
from imblearn.over_sampling import SMOTE
```

```python
oversample = SMOTE()
X, y = oversample.fit_resample(df[['f1', 'f2']], df['target'])
```
