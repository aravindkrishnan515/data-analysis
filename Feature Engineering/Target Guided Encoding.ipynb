# Target Guided Encoding

#### It's a technique used to encode the categorical variable based on the relationship with the target variable. It's used when we have large of unique categories

#### In Target Guided Encoding we replace the categorical variable with the numerical value based on the mean, median of the target variable for that category. It creates the monotanic relationship between the categorical variable and the target variable which can increase the predictive power of the model.

```python
import pandas as pd

df = pd.DataFrame({
    'city' : ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'paris'],
    'price': [200, 150, 300, 250, 180, 320]
})
```

```python
df.head()
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
import seaborn as sns
dataset = sns.load_dataset('tips')
```

```python
dataset
```

```python
bill_mean_price = dataset.groupby('time')['total_bill'].mean().to_dict()
```

```python
bill_mean_price
```

```python
dataset['time_encoded'] = dataset['time'].map(bill_mean_price)
```
