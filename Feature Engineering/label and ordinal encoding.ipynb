#### Label and Ordinal encoding are used to convert categorical data into numerical data  

# Label Encoding

#### Label encoding is converting categorical data into a numerical data without any order

#### Red : 2
#### Green : 1
#### Blue : 0

```python
import pandas as pd

df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'green', 'blue', 'red', 'red', 'red', 'green']
})
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# create a instance of LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
color_encoded = lbl_encoder.fit_transform(df[['color']])
```

```python
color_encoded
```

```python
lbl_encoder.transform(['blue'])[0]
```

# Ordinal Encoding

#### Ordinal encoding also converts categorical data into a numerical data but intentionally preserves a meaningful order

#### small : 0
#### mediaum : 1
#### large : 2

```python
df = pd.DataFrame({
    'order' : ['small', 'medium', 'small', 'large', 'medium', 'medium', 'large', 'large'],
})
```

```python
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder

# create a instance of OrdinalEncoder
encoder = OrdinalEncoder(categories = [['small', 'medium', 'large']])
```

```python
encoded = encoder.fit_transform(df[['order']])
```

```python
encoded
```

```python
encoder.transform([['small']])
```
